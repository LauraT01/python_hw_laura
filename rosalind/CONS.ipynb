```python
def read_file_into_lines(file_path):
    lines = []
    with open(file_path, 'r') as infile:
        for line in infile.readlines():
            cleaned = line.rstrip()
            lines.append(cleaned)
    return lines

def parse_fasta(lines_list):
    current_sequence = ''
    current_id = ''
    sequences = {}

    for line in lines_list:
        if line[0] == '>':
            sequences[current_id] = current_sequence
            current_id = line[1:]
            current_sequence = ''
        else:
            current_sequence = current_sequence + line

    sequences[current_id] = current_sequence
    del sequences['']
    return sequences

fasta = read_file_into_lines('/Users/laura/Desktop/python_hw/data/Rosalind_cons_sample.txt')

sequences = parse_fasta(fasta)
```

```python
matrix = []
for key in sequences:
    matrix.append(sequences[key])

A = []
T = []
C = []
G = []
n = len(matrix[0])

for sequence in matrix:
    for i in range(0, n):
        posA = 0
        posT = 0
        posC = 0
        posG = 0
        for position in sequence:
            if position == 'A':
                posA = posA + 1
            elif position == 'T':
                posT = posT + 1
            elif position == 'C':
                posC = posC + 1
            elif position == 'G':
                posG = posG + 1
    A.append(posA)
    T.append(posT)
    C.append(posC)
    G.append(posG)


```
