## Consensus and Profile

**Given**: A collection of at most 10 DNA strings of equal length (at most 1 kbp) in FASTA format.

**Return**: A consensus string and profile matrix for the collection. (If several possible consensus strings exist, then you may return any one of them.)

```python
def read_file_into_lines(file_path):
    lines = []
    with open(file_path, 'r') as infile:
        for line in infile.readlines():
            cleaned = line.rstrip()
            lines.append(cleaned)
    return lines

def parse_fasta(lines_list):
    current_sequence = ''
    current_id = ''
    sequences = {}

    for line in lines_list:
        if line[0] == '>':
            sequences[current_id] = current_sequence
            current_id = line[1:]
            current_sequence = ''
        else:
            current_sequence = current_sequence + line

    sequences[current_id] = current_sequence
    del sequences['']
    return sequences

fasta = read_file_into_lines('/Users/laura/Desktop/python_hw/data/Rosalind_cons_sample.txt')

sequences = parse_fasta(fasta)
```

```python
matrix = []
for key in sequences:
    matrix.append(sequences[key])

n = len(matrix[0])
A = [0] * n
T = [0] * n
C = [0] * n
G = [0] * n

for sequence in matrix:
    for i in range(n):
        if sequence[i] == 'A':
            A[i] += 1
        elif sequence[i] == 'T':
            T[i] += 1
        elif sequence[i] == 'C':
            C[i] += 1
        elif sequence[i] == 'G':
            G[i] += 1

```

```python
consensus = ""
for i in range(n):
    counts = {'A': A[i],
              'T': T[i],
              'C': C[i],
              'G': G[i]}
    best_fit = 'A'
    for base in ['C', 'G', 'T']:
        if counts[base] > counts[best_fit]:
            best_fit = base
    consensus += best_fit
```

```python
#unpacking operator *
print('A:', *A)
print('T:', *T)
print('C:', *C)
print('G:', *G)
print(consensus)
```
