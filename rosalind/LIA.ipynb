## Rosalind: Independent Alleles (LIA)
Given: Two positive integers k (k≤7) and N (N≤2k). We begin with Tom, who in the 0th generation has genotype Aa Bb. Tom has two children in the 1st generation, each of whom has two children, and so on. Each organism always mates with an organism having genotype Aa Bb.

Return: The probability that **at least N** Aa Bb organisms will belong to the k-th generation of Tom's family tree (don't count the Aa Bb mates at each level). **Assume that Mendel's second law holds for the factors**.

*at least N* = all events resulting in >= N

*Mendel's 2nd law* = different alleles are inherited indepentendly

either success (AaBb) or failure (AABB, AABb, AAbb, AaBB, Aabb, aaBB, aaBb, aabb) -> binomial probability

### Test data

```python
from math import factorial as f

k = 2
N = 1

kids = 2 ** k
#probability for Aa/Bb is 50%, respectively -> 0.5 * 0.5 = 0.25
p = 0.25

def factor(n, r):
    return f(n) / (f(r) * f(n - r))

```

```python
prob = 0.0
for gen in range(N, (kids +1)):
    case = factor(kids, gen) * (p ** gen) * ((1 - p) ** (kids - gen))
    prob = prob + case
```
