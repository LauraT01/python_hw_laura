### Part 1


```python
def read_file_into_lines(file_path):
    lines = []
    with open(file_path, 'r') as infile:
        for line in infile.readlines():
            cleaned = line.rstrip()
            lines.append(cleaned)
    return lines

input_lines = read_file_into_lines('/Users/laura/Desktop/python_hw/aoc25/input_day9')

floor = []
for item in input_lines:
    floor.append(item.split(','))

rectangles = set()
for red_tile in floor:
    position = int(red_tile[0])
    line = int(red_tile[1])
    for n, number in enumerate(floor):
        item = floor[n]
        next_position = int(item[0])

        x = position - next_position
        if x < 0:
            x = x * (-1)

        line_tiles = x + 1
        next_line = int(item[1])
        y = line - next_line
        if y < 0:
            y = y * (-1)
        line_counts = y + 1
        rectangle = line_tiles * line_counts
        rectangles.add(rectangle)

result = max(rectangles)
print(result)
```
