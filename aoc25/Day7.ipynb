```python
def read_file_into_lines(file_path):
    lines = []
    with open(file_path, 'r') as infile:
        for line in infile.readlines():
            cleaned = line.rstrip()
            lines.append(cleaned)
    return lines
```

```python
incoming = [manifold[0].index('S')]
total = 0
acitvated_splitters = set()

for l, line in enumerate(manifold):
    outgoing = set()
    for ray in incoming:
        if ray < 0 or ray > len(line):
            continue 
        if line[ray] == '^':
            acitvated_splitters.add((l, ray))
            split_light1 = (ray +1)
            outgoing.add(split_light1)
            split_light2 = (ray -1)
            outgoing.add(split_light2)
        else:
            outgoing.add(ray)
        
    

        
    incoming = outgoing
    
            
        
```
